```python
import logging
import numpy as np
import capytaine as cpt
import pygmsh
import gmsh
import vtk
import matplotlib.pyplot as plt
import os
import xarray as xr
import capytaine.io.mesh_writers

logging.basicConfig(level=logging.INFO,format="%(levelnames)s:\t%(message)s")
```

```python

# 25m diameter 5m height 

cylinder_mesh = cpt.mesh_vertical_cylinder(
    length=5, radius=12.5,  # Dimensions
    center=(0, 0, 0),         # Position
    resolution=(10, 50, 10)    # Fineness of the mesh
    )
body = cpt.FloatingBody(cylinder_mesh, dofs=cpt.rigid_body_dofs())

float = cpt.FloatingBody(mesh=cylinder_mesh)
float.name = "cylinder"

float.show_matplotlib()
float.show()

float_full_mesh = float.mesh

capytaine.io.mesh_writers.write_STL('cylinder.stl', float_full_mesh.vertices, float_full_mesh.faces)
```
